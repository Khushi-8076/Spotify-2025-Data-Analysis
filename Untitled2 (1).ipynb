# 7.Visualization

### Histogram

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Plot distribution for daily_streams
sns.histplot(df["Daily_streams"].dropna(), kde=True, ax=axes[0], color="skyblue")
axes[0].set_title("Distribution of Daily Streams")
axes[0].set_xlabel("Daily Streams")
axes[0].set_ylabel("Frequency")

# Plot distribution for spotify_streams_total
sns.histplot(
    df["Spotify_streams_total"].dropna(), kde=True, ax=axes[1], color="salmon"
)
axes[1].set_title("Distribution of Total Spotify Streams")
axes[1].set_xlabel("Total Spotify Streams")
axes[1].set_ylabel("Frequency")

plt.tight_layout()
plt.show()
```

### Scatter Plot

```python
plt.figure(figsize=(10,6))

sns.scatterplot(
    data=df,
    x="Spotify_streams_total",
    y="Daily_streams",
    hue="Collaboration",
    palette="Set2",
    alpha=0.8
)

plt.title("Spotify Total Streams vs Daily Streams")
plt.xlabel("Spotify Total Streams")
plt.ylabel("Daily Streams")
plt.legend(title="Collaboration")
plt.show()
```

## Bar chart visualization for Top 10 Artists

```python
plt.figure(figsize=(12, 6))
sns.barplot(
    data=df.head(10),
    x="Spotify_streams_total",
    y="Artist",
    palette="viridis",
)
plt.title("Top 10 Artists by Total Spotify Streams")
plt.xlabel("Total Stream Count")
plt.ylabel("Artist")
plt.show()
```

## Bar chart visualization for Top 10 Songs

```python
plt.figure(figsize=(12, 6))
sns.barplot(
    data=df.head(),
    x="Spotify_streams_total",
    y="track",
    hue="track",
    palette="magma",
    legend=False
)
plt.title("Top 10 Songs by Total Spotify Streams")
plt.xlabel("Total Spotify Stream")
plt.ylabel("Songs Name")
plt.show()
```

## Box Plot

```python
plt.figure(figsize=(12,6))

sns.boxplot(
    data=df,
    x="Billed_artist_count",
    y="Daily_streams_rank",
    palette="viridis"
)

plt.title("Daily Streams Rank by Billed Artist Count")
plt.xlabel("Billed Artist Count")
plt.ylabel("Daily Streams Rank")
plt.show()
```

## Count Plot

 Collaboration vs Solo Songs

```python
plt.figure(figsize=(6,5))

sns.countplot(
    data=df,
    x="Collaboration",
    hue="Collaboration",
    palette="Set2",
    legend=False
)

plt.title("Number of Collaboration vs Solo Songs")
plt.xlabel("Collaboration")
plt.ylabel("Number of Songs")

plt.show()
```

## Pie Chart

```python
collab = df["Collaboration"].value_counts()

plt.figure(figsize=(6,6))

plt.pie(
    collab,
    labels=collab.index,
    autopct="%1.1f%%",
    startangle=90,
    explode=(0.05, 0),
    shadow=True
)

plt.title("Percentage of Collaboration vs Solo Songs")

plt.show()
```

## Heatmap

Correlation Matrix

```python
plt.figure(figsize=(8,6))

corr = df[
    [
        "rank",
        "Spotify_streams_total",
        "Daily_streams",
        "Daily_stream_share_pct"
    ]
].corr()

sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5
)

plt.title("Correlation Matrix")
plt.show()
```

## Pair Plot

Relationship Between Numerical Variables

```python
sns.pairplot(
    df[
        [
            "Spotify_streams_total",
            "Daily_streams",
            "Daily_streams_rank",
            "Daily_stream_share_pct"
        ]
    ]
)

plt.show()
```

## Feature Engineering

Create a new feature

```python
df["Streams_per_artist"] = (
    df["Spotify_streams_total"] /
    df["Billed_artist_count"]
)
```

```python
df["Popularity_Index"] = (
    df["Spotify_streams_total"] *
    df["Daily_stream_share_pct"]
)
```

# Key Findings

1. The dataset contains 730 songs and 10 features, including streaming metrics, artist details, and collaboration information.

2. A small number of artists dominate Spotify streaming, contributing a large share of the total streams.

3. Songs with higher total Spotify streams generally receive higher daily streams, indicating a positive relationship between long-term and current popularity.

4. Most songs in the dataset are solo tracks, while collaboration tracks represent a smaller proportion.

5. Collaboration tracks can achieve high streaming numbers, but collaborations do not always result in better daily stream rankings.

6. The distribution of daily streams is highly skewed, with only a few songs receiving exceptionally high daily streams.

7. The correlation analysis shows a positive relationship between Spotify total streams and daily streams, while other numerical variables have weaker correlations.

8. The Top 10 Songs and Top 10 Artists visualizations highlight the most successful performers based on Spotify streaming data.

# Outliers

 1. A few songs have exceptionally high total Spotify streams, making them clear outliers in the dataset.

2. Some older songs have accumulated very high total streams but currently receive lower daily streams, indicating long-term popularity.

3. Some newly released or viral songs show very high daily streams despite having lower cumulative total streams.

4. A small number of artists dominate the streaming charts by appearing multiple times among the top-streamed songs.

5. Songs with multiple collaborating artists sometimes achieve high streaming numbers, but collaboration alone does not guarantee higher rankings.
